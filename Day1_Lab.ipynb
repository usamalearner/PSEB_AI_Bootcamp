**1A.4** Write `normalize_prompt(text, max_words=10)` that:
1. strips whitespace and lowercases,
2. collapses multiple spaces into one,
3. keeps only the first `max_words` words.

````python
def normalize_prompt(text, max_words=10):
    # 1. Strip whitespace and convert to lowercase
    text = text.strip().lower()

    # 2. Split into words (automatically collapses multiple spaces)
    words = text.split()

    # 3. Keep only the first max_words words
    words = words[:max_words]

    # Join words with a single space
    return " ".join(words)
````

````python
assert normalize_prompt("   Hello    WORLD  ") == "hello world"
assert normalize_prompt("one two three four", max_words=2) == "one two"
print("✅ 1A.4 passed")
````

## Exercise 1A (Stretch)

**1A.5** Write `is_palindrome(text)` that ignores case, spaces and punctuation.
`"A man, a plan, a canal: Panama"` is a palindrome.

````python
def is_palindrome(text):
    # Convert everything to lowercase
    text = text.lower()

    # Keep only letters and numbers
    cleaned = ""

    for char in text:
        if char.isalnum():
            cleaned += char

    # Reverse the cleaned string and compare
    return cleaned == cleaned[::-1]
````

````python
assert is_palindrome("A man, a plan, a canal: Panama")
assert is_palindrome("racecar")
assert not is_palindrome("python")
print("✅ 1A.5 passed")
````

## Demo: why sets are fast (Big-O intuition)

````python
import timeit
big_list = list(range(1_000_000))
big_set = set(big_list)

t_list = timeit.timeit(lambda: 999_999 in big_list, number=100)
t_set = timeit.timeit(lambda: 999_999 in big_set, number=100)
print(f"list: {t_list:.4f}s   set: {t_set:.6f}s   set is ~{t_list / t_set:,.0f}x faster")
````

````python
import timeit

big_list = list(range(1_000_000))
big_set = set(big_list)

for target in [0, 500_000, 999_999]:

    list_time = timeit.timeit(
        lambda: target in big_list,
        number=100
    )

    set_time = timeit.timeit(
        lambda: target in big_set,
        number=100
    )

    print(
        f"Target: {target:>7} | "
        f"List: {list_time:.6f}s | "
        f"Set: {set_time:.6f}s"
    )
````

# Lab 1C: Object-Oriented Programming

A **class** is a blueprint. An **object** (instance) is one thing built from it.
`self` is the specific object a method is working on.

## Demo: BankAccount

````python
class BankAccount:
    bank_name = "AI Bank"                 # class attribute (shared)

    def __init__(self, owner, balance=0):
        self.owner = owner                # instance attributes
        self._balance = balance           # _ means "internal, don't touch directly"

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self._balance += amount

    def withdraw(self, amount):
        if amount > self._balance:
            raise ValueError("Insufficient funds")
        self._balance -= amount

    def __str__(self):
        return f"{self.owner}: Rs {self._balance:,}"

    def __repr__(self):
        return f"BankAccount(owner={self.owner!r}, balance={self._balance})"

acc = BankAccount("Sadia", 1000)
acc.deposit(500)
acc.withdraw(200)
print(acc)
print(repr(acc))
print(acc.balance, BankAccount.bank_name)
````

````python
class Student:

    # Class attribute
    school_name = "ABC School"

    # Constructor
    def __init__(self, name, roll_no, marks=0):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    # Method to add marks
    def add_marks(self, amount):
        self.marks += amount

    # Method to deduct marks
    def deduct_marks(self, amount):
        self.marks -= amount

    # Method to display student information
    def display_info(self):
        print("Name:", self.name)
        print("Roll Number:", self.roll_no)
        print("Marks:", self.marks)
        print("School:", self.school_name)
        print("----------------------")

    # Bonus method
    def check_result(self):
        if self.marks >= 50:
            print("Result: Pass")
        else:
            print("Result: Fail")


# Create three student objects
s1 = Student("Sara", 101, 70)
s2 = Student("Ali", 102, 85)
s3 = Student("Ahmed", 103, 60)


# Perform operations
s1.add_marks(10)
s2.deduct_marks(5)
s3.add_marks(15)


# Display student information
s1.display_info()
s2.display_info()
s3.display_info()


# Check results
print("Student Results:")
s1.check_result()
s2.check_result()
s3.check_result()
````

## Demo: inheritance and polymorphism

````python
class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0, rate=0.05):
        super().__init__(owner, balance)
        self.rate = rate

    def add_interest(self):
        self.deposit(self._balance * self.rate)

    def __str__(self):                   # override
        return "[Savings] " + super().__str__()

s = SavingsAccount("Bilal", 10_000)
s.add_interest()
for account in [acc, s]:                # same method name, different behaviour
    print(account)
print(isinstance(s, BankAccount))
````

## Demo: dataclasses (less boilerplate)

````python
from dataclasses import dataclass, field

@dataclass
class Message:
    role: str
    content: str

@dataclass
class Conversation:
    messages: list = field(default_factory=list)

    def add(self, role, content):
        self.messages.append(Message(role, content))

chat = Conversation()
chat.add("user", "What is OOP?")
chat.add("assistant", "Bundling data with behaviour.")
print(chat)
print(Message("user", "hi") == Message("user", "hi"))   # __eq__ for free
````

## Demo: exceptions

Read tracebacks **from the bottom up**: the last line tells you the error type and message.

````python
def safe_int(text):
    try:
        value = int(text)
    except ValueError:
        print(f"Could not convert {text!r}")
        return None
    else:
        print("converted fine")
        return value
    finally:
        print("finally always runs")

print(safe_int("42"))
print(safe_int("forty-two"))
````

````python
class InsufficientFundsError(Exception):
    """Raised when a withdrawal is larger than the balance."""

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(f"Need {amount}, have {balance}")
    return balance - amount

try:
    withdraw(100, 500)
except InsufficientFundsError as e:
    print("Caught:", e)
````
